# PROGRAMING PROBLEMS
To access download and load the cars.csv file.

```python
import pandas as pd
```

```python
cars = pd.read_csv("C:/Users/johnn/Downloads/cars.csv")
cars.head()
```

##

## A. Positional and Label-Based Slicing

```python
print("Part 1: Display Shape and Column Names:\n")
print("Shape of cars:", cars.shape)
print("\nColumns in cars:", cars.columns.tolist())
```

```python
cars_6_to_10 = cars.iloc[5:10]
print("Part 2: Create cars Rows 6 to 10 \n")
display(cars_6_to_10)
```

```python
columns_subset = ["Model", "mpg", "cyl", "hp", "gear"]
cars_6_to_10_display = cars_6_to_10.loc[:, columns_subset]
print("Part 3: Selected Columns from cars 6 to 10 \n")
display(cars_6_to_10_display)
```

## 

## B. Model Lookup

```python
toyota = cars.loc[cars["Model"] == "Toyota Corolla"]
print("Part 1: Toyota Corolla \n")
display(toyota)
```

```python
pontiac_cols = ["Model", "mpg", "hp", "wt"]
pontiac = cars.loc[cars["Model"] == "Pontiac Firebird", pontiac_cols]
print("Part 2: Pontiac Firebird \n")
display(pontiac)
```

## 

## C. Multi-Model Subsetting

```python
target_models = ["Datsun 710", "Lotus Europa", "Ferrari Dino"]
target_columns = ["Model", "mpg", "cyl", "hp", "gear"]
selected_cars = cars.loc[cars["Model"].isin(target_models), target_columns]

print("Part 1: selected_cars \n")
display(selected_cars)
print("\nShape of selected_cars:", selected_cars.shape)
```
